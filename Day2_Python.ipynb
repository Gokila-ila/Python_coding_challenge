#Tasks:
####Ask the user to input a sentence (representing feedback).


```python
# Feedback Analyzer
feedback=input("Please Enter the feedback : ")
```

####Count how many words are in the sentence


```python
#count no of words
word_count=len(feedback.split())
print("Number of words: ",word_count)

```

####      Check how many times the word “data” appears in the sentence, regardless of case.


```python
#Count 'data' word present
data_count=feedback.lower().count('data')
print("The word data appears:",data_count)
```

####Extract and print the first 5 and last 5 characters of the feedback separately.


```python
#Print first 5 character
first5=feedback[:5]
print("First Five Character :",first5)
```

```python
#print last 5 character
last5=feedback[-5]
print("Last Five characyers are: ",last5)
```

####Reverse the sentence and print it.


```python
#Reverse sentence
reverse=feedback[::-1]
print("Reverse sentences ",reverse)
```
